```python
# Start coding here
# Use as many cells as you need
def validate_user(name,email,password):
    if validate_email(email) and validate_name(name) and validate_password(password):
        return True
    else:
        raise ValueError("Validation failed.")
            
```

```python
def register_user(name,email,password):
    try:
        validate_user(name,email,password)
        dictionary = {"name":name,"email":email, "password":password} 
        return dictionary
    except ValueError:
        return False
register_user("resul","resul@hotmail.com","Resuloooo1")
register_user("a","b","ab")
```
